## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">Instantiate and inspect the raw environment</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;"><code>make_sepsis_env()</code> builds the discrete <code>ICU-Sepsis-v2</code> MDP and applies the fixed difficulty (<code>SOFA_BIAS = 5.0</code>, <code>LAM = 0.02</code>). We inspect its observation and action spaces, then extract the full transition and reward tensors <code>P</code> and <code>R</code> — Config A's defining affordance, which makes both Bellman backups and analytical policy evaluation possible later.</p>
</div>

```python
env = make_sepsis_env()
obs, info = env.reset(seed=SEED)

print(f'Observation space : {env.observation_space} discrete integer state')
print(f'Action space      : {env.action_space}')
print(f'Initial state     : {obs}')
print()

raw = env.unwrapped
P = raw._tx_mat
R_sasp = raw._r_mat
R = (P * R_sasp).sum(axis=2)

print(f'Transition matrix P : {P.shape}  (S × A × S\')')
print(f'Reward matrix R     : {R.shape}  (S × A)')
print(f'Reward range        : [{R.min():.3f}, {R.max():.3f}]')
print()
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">Random baseline: the performance floor</div>

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">State visitation and reward structure</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Two diagnostics about the MDP: which states the agent encounters under a random policy, and where the reward signal lives. Together they motivate the sparse-reward, short-horizon framing used by every Config A algorithm.</p>
</div>

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

np.random.seed(SEED)
env_vis = make_sepsis_env(quiet=True)
env_vis.action_space.seed(SEED)
visited = []
for _ in range(300):
    obs, _ = env_vis.reset(seed=np.random.randint(100_000))
    done = False
    while not done:
        visited.append(int(obs))
        obs, _, te, tr, _ = env_vis.step(env_vis.action_space.sample())
        done = te or tr
env_vis.close()
clinical = [s for s in visited if s not in (STATE_SURVIVED, STATE_DIED)]

axes[0].hist(clinical, bins=60, color='steelblue', edgecolor='none', alpha=0.8)
axes[0].set_xlabel('State index (0–713 = clinical)')
axes[0].set_ylabel('Visit count')
axes[0].set_title('State visitation random policy (300 episodes)', fontweight='bold')

axes[1].bar(range(N_STATES), R.max(axis=1), color='tomato', width=1.0, alpha=0.8)
axes[1].set_xlabel('State index')
axes[1].set_ylabel('max_a R(s,a)')
axes[1].set_title('Reward sparsity only terminal transitions carry reward',
                  fontweight='bold')

plt.tight_layout()
plt.savefig(f'{PLOTS_DIR}/configA_env_exploration.png', bbox_inches='tight')
plt.show()
```

    <h3 style="margin-top: 0; margin-bottom: 10px; color: #333333; font-weight: bold;">Why tabular methods</h3>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">With 716 discrete states and 25 actions, the Q-table has shape (716, 25) — 17,900 entries. That size is manageable in memory, which is exactly what motivates tabular algorithms here.</p>

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">A.1 Value Iteration (model-based optimal reference)</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Config A exposes the full MDP (the transition tensor <code>P</code> and expected-reward matrix <code>R</code> extracted above), so the optimal policy can be computed directly. Value Iteration repeatedly applies the Bellman optimality backup, updating each state's value to the best one-step return achievable under the known model, and stops once the value function settles. Because it uses the true model, the resulting policy <code>pi_VI</code> is the optimal reference against which every model-free method in A.2 and A.3 is measured.</p>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">We evaluate <code>pi_VI</code> with the same 1000-episode protocol and fixed seeding as the random baseline, so its Monte-Carlo return is directly comparable to the other algorithms.</p>
</div>

```python
V = np.zeros(N_STATES)
vi_deltas = []
vi_policy_changes = []
pi_prev = -np.ones(N_STATES, dtype=int)
for it in range(1000):
    Q_sa = R + GAMMA * P.dot(V)
    V_new = Q_sa.max(axis=1)
    pi_k = Q_sa.argmax(axis=1)
    delta = float(np.max(np.abs(V_new - V)))
    if it == 0:
        n_changed = N_STATES
    else:
        policy_change_mask = pi_k != pi_prev
        n_changed = int(policy_change_mask.sum())
    vi_deltas.append(delta)
    vi_policy_changes.append(n_changed)
    pi_prev = pi_k
    V = V_new
    if delta < 1e-10:
        break

V_VI = V
Q_VI = Q_sa
pi_VI = pi_k
vi_iters = len(vi_deltas)

vi_returns, vi_lengths = evaluate_policy(pi_VI)
vi_survived = vi_returns > 0
vi_mean_ret = float(vi_returns.mean())
vi_surv = float(vi_survived.mean()) * 100
vi_mean_len = float(vi_lengths.mean())

rand_survived = rand_returns > 0
rand_mean = float(rand_returns.mean())
rand_surv = float(rand_survived.mean()) * 100

print(f'Value Iteration converged in {vi_iters} iterations (final Δ = {vi_deltas[-1]:.2e})')
print(f'Mean return    : {vi_mean_ret:.4f}    (random baseline: {rand_mean:.4f})')
print(f'Survival rate  : {vi_surv:.1f}%        (random baseline: {rand_surv:.1f}%)')
print(f'Mean ep length : {vi_mean_len:.1f} steps')
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">Convergence diagnostics</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Two views of how Value Iteration converges. The Bellman-residual panel (log scale) tracks the largest change in any state's value between iterations, decaying geometrically toward the 1e-10 tolerance. The policy-stability panel counts how many states change their greedy action each iteration. The policy usually locks in long before the value function fully converges, so the final iterations are numerical polish with no effect on the decisions made.</p>
</div>

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

iters_axis = np.arange(1, vi_iters + 1)
vi_deltas_arr = np.array(vi_deltas)
vi_policy_changes_arr = np.array(vi_policy_changes)

post_init_zero_mask = vi_policy_changes_arr[1:] == 0
if post_init_zero_mask.any():
    first_zero_slice_idx = int(np.argmax(post_init_zero_mask))
    iter_policy_stable = first_zero_slice_idx + 2
else:
    iter_policy_stable = vi_iters

axes[0].semilogy(iters_axis, vi_deltas_arr, color='darkgreen', linewidth=1.8)
axes[0].axhline(1e-10, color='gray', linestyle='--', linewidth=1.0,
                label='Convergence threshold (1e-10)')
axes[0].axvline(vi_iters, color='tomato', linestyle=':', linewidth=1.2,
                label=f'Converged at iter {vi_iters}')
axes[0].set_xlabel('Iteration')
axes[0].set_ylabel(r'$\|V_{k+1} - V_k\|_\infty$  (log scale)')
axes[0].set_title('Bellman residual', fontweight='bold')
axes[0].legend(loc='upper right')
axes[0].grid(True, which='both', alpha=0.3)

policy_panel_title = f'Policy stable at iter {iter_policy_stable} < value convergence at {vi_iters}'
axes[1].plot(iters_axis, vi_policy_changes_arr, color='tomato', linewidth=1.8)
axes[1].axvline(iter_policy_stable, color='gray', linestyle=':', linewidth=1.2,
                label=f'Stable at iter {iter_policy_stable}')
axes[1].set_xlabel('Iteration')
axes[1].set_ylabel('# states with policy change')
axes[1].set_title(policy_panel_title, fontweight='bold')
axes[1].legend(loc='upper right')
axes[1].grid(True, alpha=0.3)

fig.suptitle('Value Iteration convergence — Config A', fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig(f'{PLOTS_DIR}/configA_vi_convergence.png', bbox_inches='tight')
plt.show()
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">Exact policy evaluation via linear solve</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Because the MDP is known, the expected return of any policy can be computed exactly without rollouts by solving a small linear system for the policy's value function over the non-terminal states. The Monte-Carlo estimates are kept too (for the return-distribution plot and as a simulator-versus-model sanity check), but this analytical return is the precise number used for the optimality-gap comparison.</p>
    <div style="background-color: #A8C5E0; padding: 10px; margin-top: 10px; margin-right: 30px; border-left: 5px solid #4682B4; border-radius: 5px; max-width: 95%; color: #333333;">
        <strong>Note:</strong> The gym wrapper ends an episode on reaching state 713 (died) or 714 (survived), but in the underlying MDP both transition onward into 715, the only true absorbing state. The linear solve therefore uses the structural absorber set (detected via self-loop probability): treating 713 and 714 as terminal would make the system singular for the random policy.
    </div>
</div>

```python
all_states = np.arange(N_STATES)
self_loop_prob = np.zeros((N_STATES, N_ACTIONS))
for s in all_states:
    self_loop_prob[s, :] = P[s, :, s]
is_absorbing_per_action = self_loop_prob == 1.0
is_absorbing = is_absorbing_per_action.all(axis=1)
NON_TERMINAL = all_states[~is_absorbing]


def analytical_J(policy=None):
    """Exact expected return d_0 · V_π.

    policy: deterministic policy (N_STATES,) action array, or None for the
        uniform-random policy.
    Returns (J, V_π) with V_π zero on terminal states.
    """
    if policy is None:
        P_pi = P.mean(axis=1)
        R_pi = R.mean(axis=1)
    else:
        state_idx = np.arange(N_STATES)
        P_pi = P[state_idx, policy, :]
        R_pi = R[state_idx, policy]

    P_nt = P_pi[np.ix_(NON_TERMINAL, NON_TERMINAL)]
    R_nt = R_pi[NON_TERMINAL]
    A_mat = np.eye(len(NON_TERMINAL)) - GAMMA * P_nt

    V_pi = np.zeros(N_STATES)
    V_pi[NON_TERMINAL] = np.linalg.solve(A_mat, R_nt)
    J = float(raw._d_0 @ V_pi)
    out = (J, V_pi)
    return out


J_rand_exact, _ = analytical_J(policy=None)
J_VI_exact, V_VI_exact = analytical_J(pi_VI)


mc_vs_analytical_gap = abs(vi_mean_ret - J_VI_exact)

print('Analytical J (exact, no sampling):')
print(f'  Random policy       : {J_rand_exact:.4f}')
print(f'  VI policy           : {J_VI_exact:.4f}')
print(f'  MC mean (1000, VI)  : {vi_mean_ret:.4f}   → simulator-model gap {mc_vs_analytical_gap:.4f}')
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">A.2 Q-Learning (model-free)</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Q-Learning is the model-free TD(0) counterpart to the Value Iteration cell above. Including it despite having <code>P</code> and <code>R</code> validates that an agent without access to the MDP model can recover essentially the same policy by interacting with the simulator. We train across 3 seeds so the learning curve can show mean ± std, and pick the median seed for downstream policy visualisations.</p>
</div>

```python
N_EPISODES = 200_000
ALPHA = 0.1
EPS_START = 1.0
EPS_END = 0.05
EPS_DECAY_END = int(0.8 * N_EPISODES)
SEEDS = [0, 1, 2]
SNAPSHOT_EVERY = 5_000

ql_eps_schedule = np.empty(N_EPISODES)
for ep in range(N_EPISODES):
    frac = min(ep / EPS_DECAY_END, 1.0)
    ql_eps_schedule[ep] = EPS_START + (EPS_END - EPS_START) * frac

snapshot_eps_list = list(range(SNAPSHOT_EVERY, N_EPISODES + 1, SNAPSHOT_EVERY))
if snapshot_eps_list[-1] != N_EPISODES:
    snapshot_eps_list.append(N_EPISODES)
snapshot_eps = np.array(snapshot_eps_list)

Q_QL_per_seed = []
ql_curves = np.empty((len(SEEDS), N_EPISODES))
ql_lengths_train = np.empty((len(SEEDS), N_EPISODES))
ql_td_errors = np.empty((len(SEEDS), N_EPISODES))
ql_snapshots_per_seed = []

for s_idx, seed in enumerate(SEEDS):
    env_train = make_sepsis_env(quiet=True)
    env_train.action_space.seed(seed)
    rng = np.random.default_rng(seed)
    Q = np.zeros((N_STATES, N_ACTIONS))
    seed_snapshots = []

    for ep in range(N_EPISODES):
        eps = ql_eps_schedule[ep]
        obs, _ = env_train.reset(seed=int(rng.integers(0, 100_000)))
        total_r = 0.0
        steps = 0
        abs_td_sum = 0.0
        done = False
        while not done:
            if rng.random() < eps:
                action = int(env_train.action_space.sample())
            else:
                action = int(Q[obs].argmax())
            next_obs, r, te, tr, _ = env_train.step(action)
            terminal = te or tr
            if terminal:
                target = r
            else:
                target = r + GAMMA * Q[next_obs].max()
            td_error = target - Q[obs, action]
            Q[obs, action] += ALPHA * td_error
            abs_td_sum += abs(td_error)
            total_r += r
            steps += 1
            obs = next_obs
            done = terminal

        ql_curves[s_idx, ep] = total_r
        ql_lengths_train[s_idx, ep] = steps
        ql_td_errors[s_idx, ep] = abs_td_sum / max(steps, 1)

        if (ep + 1) in snapshot_eps_list:
            seed_snapshots.append(Q.copy())

    env_train.close()
    Q_QL_per_seed.append(Q)
    ql_snapshots_per_seed.append(seed_snapshots)

ql_returns_per_seed = []
ql_lengths_per_seed = []
for Q in Q_QL_per_seed:
    pi_seed = Q.argmax(axis=1)
    seed_returns, seed_lengths = evaluate_policy(pi_seed)
    ql_returns_per_seed.append(seed_returns)
    ql_lengths_per_seed.append(seed_lengths)

seed_means_list = []
seed_survs_list = []
seed_lens_list = []
for r_arr, l_arr in zip(ql_returns_per_seed, ql_lengths_per_seed):
    seed_means_list.append(r_arr.mean())
    survival_pct = float((r_arr > 0).mean()) * 100
    seed_survs_list.append(survival_pct)
    seed_lens_list.append(l_arr.mean())
ql_means = np.array(seed_means_list)
ql_survs = np.array(seed_survs_list)
ql_lens = np.array(seed_lens_list)

median_value = float(np.median(ql_means))
distances = np.abs(ql_means - median_value)
median_seed_idx = int(distances.argmin())
Q_QL = Q_QL_per_seed[median_seed_idx]
pi_QL = Q_QL.argmax(axis=1)

ql_pooled_returns = np.concatenate(ql_returns_per_seed)
ql_pooled_lengths = np.concatenate(ql_lengths_per_seed)
ql_mean_ret = float(ql_pooled_returns.mean())
ql_surv = float((ql_pooled_returns > 0).mean()) * 100
ql_mean_len = float(ql_pooled_lengths.mean())

print(f'Q-Learning trained for {N_EPISODES} episodes × {len(SEEDS)} seeds (α={ALPHA}, ε: {EPS_START}→{EPS_END})')
print(f'  Per-seed mean return   : {np.array2string(ql_means, precision=4)}')
print(f'  Per-seed survival rate : {np.array2string(ql_survs, precision=1)}%')
print(f'  Pooled mean return     : {ql_mean_ret:.4f}    (VI: {vi_mean_ret:.4f}, random: {rand_mean:.4f})')
print(f'  Pooled survival rate   : {ql_surv:.1f}%        (VI: {vi_surv:.1f}%, random: {rand_surv:.1f}%)')
print(f'  Pooled mean ep length  : {ql_mean_len:.1f} steps')
print(f'  Median seed for policy viz : seed={SEEDS[median_seed_idx]} (mean eval return {ql_means[median_seed_idx]:.4f})')
print(f'  Snapshots per seed     : {len(snapshot_eps)} (every {SNAPSHOT_EVERY} episodes)')
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">Q-Learning learning curve</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Rolling-500-episode mean return averaged across seeds with a ±1 std band, and the ε schedule overlaid on a twin axis. The horizontal dashed line marks the Value Iteration ceiling.</p>
</div>

```python
WINDOW = 500

rolling_per_seed = []
for s_idx in range(len(SEEDS)):
    seed_curve = ql_curves[s_idx]
    seed_rolling = rolling_mean(seed_curve, WINDOW)
    rolling_per_seed.append(seed_rolling)
rolling_stack = np.vstack(rolling_per_seed)
rolling_mean_across_seeds = rolling_stack.mean(axis=0)
rolling_std_across_seeds = rolling_stack.std(axis=0)
x_axis = np.arange(WINDOW - 1, N_EPISODES)

fig, ax1 = plt.subplots(figsize=(12, 5))

ax1.plot(x_axis, rolling_mean_across_seeds, color='steelblue', linewidth=2.0,
         label=f'Q-Learning mean across {len(SEEDS)} seeds (rolling-{WINDOW})')
upper_band = rolling_mean_across_seeds + rolling_std_across_seeds
lower_band = rolling_mean_across_seeds - rolling_std_across_seeds
ax1.fill_between(x_axis, lower_band, upper_band, color='steelblue', alpha=0.2,
                 label='±1 std across seeds')

ax1.axhline(vi_mean_ret, color='darkgreen', linestyle='--', linewidth=1.5,
            label=f'Value Iteration ({vi_mean_ret:.3f})')
ax1.axhline(rand_mean, color='gray', linestyle=':', linewidth=1.5,
            label=f'Random baseline ({rand_mean:.3f})')

ax1.set_xlabel('Training episode')
ax1.set_ylabel('Episode return (rolling mean)')
ax1.set_title('Q-Learning learning curve — Config A', fontweight='bold')
ax1.legend(loc='lower right')
ax1.grid(True, alpha=0.3)

ax2 = ax1.twinx()
ax2.plot(np.arange(N_EPISODES), ql_eps_schedule, color='tomato', linewidth=1.2,
         alpha=0.7, label='ε (exploration rate)')
ax2.set_ylabel('ε', color='tomato')
ax2.tick_params(axis='y', labelcolor='tomato')
ax2.set_ylim(0, 1.05)
ax2.legend(loc='upper right')

plt.tight_layout()
plt.savefig(f'{PLOTS_DIR}/configA_qlearning_curve.png', bbox_inches='tight')
plt.show()
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">A.3 SARSA (on-policy TD)</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">SARSA is the on-policy counterpart to Q-Learning. They differ in a single line of the update: Q-Learning bootstraps from max over a' of Q(s', a') (the value of the greedy next action), while SARSA bootstraps from Q(s', a') for the action a' actually sampled under the current ε-greedy policy.</p>
        <p style="margin: 10px 0; margin-right: 40px; color: #757575;">In a clinical-safety frame this difference is worth flagging: Q-Learning&#39;s value estimate ignores the exploration steps it will continue to take, while SARSA&#39;s tracks them. Whether that translates into a safer policy in this MDP is an empirical question, settled in A.4 once both algorithms are trained.</p>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">We mirror the Q-Learning setup exactly (200k episodes × 3 seeds, α = 0.1, ε decay 1.0 → 0.05 over 80% of training, identical 1000-episode evaluation set) so the two algorithms are directly comparable in the A.4 table and the optimality-gap plot.</p>
</div>

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">SARSA training across 3 seeds</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Same hyperparameters and ε schedule as Q-Learning; the only change is the bootstrap target. Q-snapshots are captured here for the optimality-gap analysis.</p>
</div>

```python
Q_SARSA_per_seed = []
sarsa_curves = np.empty((len(SEEDS), N_EPISODES))
sarsa_lengths_train = np.empty((len(SEEDS), N_EPISODES))
sarsa_td_errors = np.empty((len(SEEDS), N_EPISODES))
sarsa_snapshots_per_seed = []


def eps_greedy_action(Q_row, eps_val, n_actions, rng_local, env_local):
    if rng_local.random() < eps_val:
        out = int(env_local.action_space.sample())
    else:
        out = int(Q_row.argmax())
    return out


for s_idx, seed in enumerate(SEEDS):
    env_train = make_sepsis_env(quiet=True)
    env_train.action_space.seed(seed)
    rng = np.random.default_rng(seed)
    Q = np.zeros((N_STATES, N_ACTIONS))
    seed_snapshots = []

    for ep in range(N_EPISODES):
        eps = ql_eps_schedule[ep]
        obs, _ = env_train.reset(seed=int(rng.integers(0, 100_000)))
        action = eps_greedy_action(Q[obs], eps, N_ACTIONS, rng, env_train)
        total_r = 0.0
        steps = 0
        abs_td_sum = 0.0
        done = False
        while not done:
            next_obs, r, te, tr, _ = env_train.step(action)
            terminal = te or tr
            if terminal:
                target = r
                next_action = 0
            else:
                next_action = eps_greedy_action(Q[next_obs], eps, N_ACTIONS, rng, env_train)
                target = r + GAMMA * Q[next_obs, next_action]
            td_error = target - Q[obs, action]
            Q[obs, action] += ALPHA * td_error
            abs_td_sum += abs(td_error)
            total_r += r
            steps += 1
            obs = next_obs
            action = next_action
            done = terminal

        sarsa_curves[s_idx, ep] = total_r
        sarsa_lengths_train[s_idx, ep] = steps
        sarsa_td_errors[s_idx, ep] = abs_td_sum / max(steps, 1)

        if (ep + 1) in snapshot_eps_list:
            seed_snapshots.append(Q.copy())

    env_train.close()
    Q_SARSA_per_seed.append(Q)
    sarsa_snapshots_per_seed.append(seed_snapshots)

sarsa_returns_per_seed = []
sarsa_lengths_per_seed = []
for Q in Q_SARSA_per_seed:
    pi_seed = Q.argmax(axis=1)
    seed_returns, seed_lengths = evaluate_policy(pi_seed)
    sarsa_returns_per_seed.append(seed_returns)
    sarsa_lengths_per_seed.append(seed_lengths)

sarsa_seed_means_list = []
sarsa_seed_survs_list = []
sarsa_seed_lens_list = []
for r_arr, l_arr in zip(sarsa_returns_per_seed, sarsa_lengths_per_seed):
    sarsa_seed_means_list.append(r_arr.mean())
    survival_pct = float((r_arr > 0).mean()) * 100
    sarsa_seed_survs_list.append(survival_pct)
    sarsa_seed_lens_list.append(l_arr.mean())
sarsa_means = np.array(sarsa_seed_means_list)
sarsa_survs = np.array(sarsa_seed_survs_list)
sarsa_lens = np.array(sarsa_seed_lens_list)

sarsa_median_value = float(np.median(sarsa_means))
sarsa_distances = np.abs(sarsa_means - sarsa_median_value)
sarsa_median_seed_idx = int(sarsa_distances.argmin())
Q_SARSA = Q_SARSA_per_seed[sarsa_median_seed_idx]
pi_SARSA = Q_SARSA.argmax(axis=1)

sarsa_pooled_returns = np.concatenate(sarsa_returns_per_seed)
sarsa_pooled_lengths = np.concatenate(sarsa_lengths_per_seed)
sarsa_mean_ret = float(sarsa_pooled_returns.mean())
sarsa_surv = float((sarsa_pooled_returns > 0).mean()) * 100
sarsa_mean_len = float(sarsa_pooled_lengths.mean())

print(f'SARSA trained for {N_EPISODES} episodes × {len(SEEDS)} seeds (α={ALPHA}, ε: {EPS_START}→{EPS_END})')
print(f'  Per-seed mean return   : {np.array2string(sarsa_means, precision=4)}')
print(f'  Per-seed survival rate : {np.array2string(sarsa_survs, precision=1)}%')
print(f'  Pooled mean return     : {sarsa_mean_ret:.4f}    (QL: {ql_mean_ret:.4f}, VI: {vi_mean_ret:.4f})')
print(f'  Pooled survival rate   : {sarsa_surv:.1f}%        (QL: {ql_surv:.1f}%, VI: {vi_surv:.1f}%)')
print(f'  Pooled mean ep length  : {sarsa_mean_len:.1f} steps')
print(f'  Median seed for policy viz : seed={SEEDS[sarsa_median_seed_idx]} (mean eval return {sarsa_means[sarsa_median_seed_idx]:.4f})')
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">Optimality gap during training</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">For each Q-table snapshot taken during training we derive the greedy policy and compute its exact analytical return J(πₜ) via the linear solve from A.1. Plotting J(πₜ)/J(π_VI) over episodes shows how quickly each model-free method approaches the model-based optimum. The dashed line at 1.0 is the Value Iteration ceiling.</p>
</div>

```python
def snapshots_to_J(snapshots_per_seed):
    n_seeds = len(snapshots_per_seed)
    n_snaps = len(snapshots_per_seed[0])
    J_arr = np.empty((n_seeds, n_snaps))
    for s_i, seed_snaps in enumerate(snapshots_per_seed):
        for k, Q_snap in enumerate(seed_snaps):
            pi_snap = Q_snap.argmax(axis=1)
            J_val, _ = analytical_J(pi_snap)
            J_arr[s_i, k] = J_val
    return J_arr


ql_J_traj = snapshots_to_J(ql_snapshots_per_seed)
sarsa_J_traj = snapshots_to_J(sarsa_snapshots_per_seed)

ql_gap = ql_J_traj / J_VI_exact
sarsa_gap = sarsa_J_traj / J_VI_exact

ql_gap_mean = ql_gap.mean(axis=0)
ql_gap_std = ql_gap.std(axis=0)
sarsa_gap_mean = sarsa_gap.mean(axis=0)
sarsa_gap_std = sarsa_gap.std(axis=0)

fig, ax = plt.subplots(figsize=(11, 5))

ax.plot(snapshot_eps, ql_gap_mean, color='steelblue', linewidth=2.0, marker='o',
        markersize=4, label='Q-Learning (mean across seeds)')
ax.fill_between(snapshot_eps, ql_gap_mean - ql_gap_std, ql_gap_mean + ql_gap_std,
                color='steelblue', alpha=0.2, label='Q-Learning ±1 std')

ax.plot(snapshot_eps, sarsa_gap_mean, color='darkorange', linewidth=2.0, marker='s',
        markersize=4, label='SARSA (mean across seeds)')
ax.fill_between(snapshot_eps, sarsa_gap_mean - sarsa_gap_std, sarsa_gap_mean + sarsa_gap_std,
                color='darkorange', alpha=0.2, label='SARSA ±1 std')

ax.axhline(1.0, color='darkgreen', linestyle='--', linewidth=1.5,
           label='Value Iteration optimum (J = 1.000)')

random_ratio = J_rand_exact / J_VI_exact
ax.axhline(random_ratio, color='gray', linestyle=':', linewidth=1.2,
           label=f'Random policy ratio ({random_ratio:.3f})')

ax.set_xlabel('Training episode')
ax.set_ylabel(r'Optimality ratio  $J(\pi_t) / J(\pi_{VI})$')
ax.set_title('Optimality gap during training — Config A', fontweight='bold')
ax.set_ylim(min(random_ratio - 0.05, 0.5), 1.05)
ax.legend(loc='lower right', fontsize=9)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(f'{PLOTS_DIR}/configA_optimality_gap.png', bbox_inches='tight')
plt.show()

print('Final optimality ratio (mean ± std across seeds):')
print(f'  Q-Learning : {ql_gap_mean[-1]:.4f} ± {ql_gap_std[-1]:.4f}')
print(f'  SARSA      : {sarsa_gap_mean[-1]:.4f} ± {sarsa_gap_std[-1]:.4f}')
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">A.6 Exploration vs exploitation</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">A.2 and A.3 trained one fixed ε-greedy schedule. To make the trade-off explicit we re-train Q-Learning under five contrasting regimes and read each one's exact policy quality via the A.1 analytical solve. The compute is light: only the schedule changes, the seed, α, and 50,000-episode budget are held identical so the curves are comparable point-by-point.</p>
    <h4 style="margin-top: 15px; margin-bottom: 8px; color: #4682B4; font-weight: bold;">Five regimes</h4>
    <ul style="margin: 10px 40px 10px 20px; padding-left: 20px; padding-right: 0; color: #757575;">
        <li style="margin-right: 20px;"><strong>Pure exploit (ε = 0)</strong> — always take the greedy action under the current Q. With Q initialised to zero this collapses to a single arbitrary action per state.</li>
        <li style="margin-right: 20px;"><strong>Pure random (ε = 1)</strong> — the random policy. The TD update still runs but the behaviour never changes; this is the floor curve.</li>
        <li style="margin-right: 20px;"><strong>Fixed ε = 0.3</strong> — heavy exploration throughout, never anneals.</li>
        <li style="margin-right: 20px;"><strong>Decay to ε = 0.05 (default)</strong> — the schedule used in A.2 and A.3.</li>
        <li style="margin-right: 20px;"><strong>Decay to ε = 0.01</strong> — same decay shape but a smaller residual exploration floor.</li>
    </ul>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Each regime's Q-table is snapshotted every 2,500 episodes; the analytical J(πₜ)/J(π_VI) ratio is computed at each snapshot. The plot below tracks both the rise and the asymptote.</p>
</div>

```python
SWEEP_EPISODES = 50_000
SWEEP_SNAPSHOT_EVERY = 2_500
SWEEP_SEED = 0
SWEEP_DECAY_END = int(0.8 * SWEEP_EPISODES)


def eps_constant(value):
    def eps_fn(ep):
        del ep
        return value
    return eps_fn


def eps_decay(eps_start, eps_end):
    def eps_fn(ep):
        frac = min(ep / SWEEP_DECAY_END, 1.0)
        out = eps_start + (eps_end - eps_start) * frac
        return out
    return eps_fn


def train_ql_with_schedule(seed, eps_fn, n_episodes=SWEEP_EPISODES, snapshot_every=SWEEP_SNAPSHOT_EVERY, alpha=ALPHA):
    env_train = make_sepsis_env(quiet=True)
    env_train.action_space.seed(seed)
    rng = np.random.default_rng(seed)
    Q = np.zeros((N_STATES, N_ACTIONS))
    snap_eps_list = list(range(snapshot_every, n_episodes + 1, snapshot_every))
    snap_J_list = []
    for ep in range(n_episodes):
        eps = eps_fn(ep)
        obs, _ = env_train.reset(seed=int(rng.integers(0, 100_000)))
        done = False
        while not done:
            if rng.random() < eps:
                action = int(env_train.action_space.sample())
            else:
                action = int(Q[obs].argmax())
            next_obs, r, te, tr, _ = env_train.step(action)
            terminal = te or tr
            if terminal:
                target = r
            else:
                target = r + GAMMA * Q[next_obs].max()
            td_error = target - Q[obs, action]
            Q[obs, action] += alpha * td_error
            obs = next_obs
            done = terminal
        if (ep + 1) in snap_eps_list:
            pi_snap = Q.argmax(axis=1)
            J_snap_value, _ = analytical_J(pi_snap)
            snap_J_list.append(J_snap_value)
    env_train.close()
    snap_eps_arr = np.array(snap_eps_list)
    snap_J_arr = np.array(snap_J_list)
    return snap_eps_arr, snap_J_arr


regime_specs = [
    ('Pure exploit (eps=0)', eps_constant(0.0), 'tab:red', '--'),
    ('Pure random (eps=1)', eps_constant(1.0), 'gray', ':'),
    ('Fixed eps=0.3', eps_constant(0.3), 'tab:olive', '-.'),
    ('Decay 1.0 -> 0.05 (default)', eps_decay(1.0, 0.05), 'steelblue', '-'),
    ('Decay 1.0 -> 0.01', eps_decay(1.0, 0.01), 'darkviolet', '-'),
]

regime_curves = {}
for label, eps_fn, color, linestyle in regime_specs:
    snap_eps_arr, snap_J_arr = train_ql_with_schedule(SWEEP_SEED, eps_fn)
    regime_curves[label] = (snap_eps_arr, snap_J_arr / J_VI_exact)
    final_ratio = regime_curves[label][1][-1]
    print(f'{label:32s}  final J/J* = {final_ratio:.3f}')

fig, ax = plt.subplots(figsize=(11, 5))
for (label, _, color, linestyle), data_tuple in zip(regime_specs, regime_curves.values()):
    snap_eps_arr, ratio = data_tuple
    ax.plot(snap_eps_arr, ratio, color=color, linewidth=2.0, linestyle=linestyle, marker='o', markersize=3, label=label)

ax.axhline(1.0, color='darkgreen', linestyle='--', linewidth=1.2, label='Value Iteration optimum')
random_ratio = J_rand_exact / J_VI_exact
ax.axhline(random_ratio, color='black', linestyle=':', linewidth=1.0, label=f'Random ratio ({random_ratio:.3f})')

ax.set_xlabel('Training episode')
ax.set_ylabel(r'Optimality ratio  $J(\pi_t) / J(\pi_{VI})$')
ax.set_title('Exploration vs exploitation - Config A Q-Learning sweep', fontweight='bold')
ax.set_ylim(min(random_ratio - 0.05, 0.4), 1.02)
ax.legend(loc='lower right', fontsize=9)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(f'{PLOTS_DIR}/configA_explore_exploit.png', bbox_inches='tight')
plt.show()
```

    <h3 style="margin-top: 0; margin-bottom: 10px; color: #333333; font-weight: bold;">Main Takeaways</h3>
    <ul style="margin: 10px 40px 10px 20px; padding-left: 20px; padding-right: 0; color: #757575;">
        <li style="margin-right: 20px;"><strong>The result inverts the textbook ranking.</strong> At 50,000 episodes the final J/J* ratios are: pure exploit (&epsilon;=0) 0.917; fixed &epsilon;=0.3 0.860; decay-to-0.01 0.834; pure random (&epsilon;=1) 0.829; decay-to-0.05 0.829. The "do nothing but learn its value" policy beats every other regime at this budget.</li>

A.4 optimality-gap plot) &mdash; still below the 50k pure-exploit number.</li>
        <li style="margin-right: 20px;"><strong>Fixed &epsilon; = 0.3 is the surprise second-place finisher.</strong> Persistent exploration prevents the agent from locking into either trap (the always-action-0 fixed point or the uniform-random plateau) and finds a policy that is 0.860 of optimal.</li>

    <h3 style="margin-top: 18px; margin-bottom: 10px; color: #333333; font-weight: bold;">Interpretation</h3>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">The exploration / exploitation lesson here is the inverse of the standard one: in MDPs where a degenerate-but-cheap policy is competitive, more exploration is not always better, and the right starting policy can outperform an &epsilon;-greedy schedule that has not yet settled. The result also explains B.1: Linear-Q converges to "always action 0" for the same reason &mdash; that policy is good enough that the algorithm has weak incentive to escape it.</p>

<div style="background: linear-gradient(to right, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C);
            padding: 10px; color: white; text-align: center; max-width: 98%;">
    <center><h1 style="margin-top: 10px; margin-bottom: 4px; color: white;
                       font-size: 32px; font-family: 'Roboto', 'Segoe UI', sans-serif;">
        <b>3. Config B — Function Approximation</b></h1></center>
</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <h3 style="margin-top: 0; margin-bottom: 10px; color: #333333; font-weight: bold;">Function approximation on the clinical environment</h3>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Config B keeps the same underlying MDP but replaces the discrete state index with the 47-dimensional continuous physiological vector and adds the three clinical-reality wrappers introduced in <strong>1. Explore the Environment</strong> (episodic observation noise, episodic missing labs, and rare acute events). A Q-table is no longer possible — there is one real-valued vector per state — so the agent must generalise across states it never sees twice. We therefore move from tabular methods to function approximation.</p>
    <h4 style="margin-top: 15px; margin-bottom: 8px; color: #4682B4; font-weight: bold;">Algorithm choice</h4>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">The two required algorithms share the Q-learning objective of Config A but differ in the approximator. Holding the learning rule fixed and changing only how Q is represented isolates a single question.</p>
    <ul style="margin: 10px 40px 10px 20px; padding-left: 20px; padding-right: 0; color: #757575;">
        <li style="margin-right: 20px;"><strong>B.1 — Linear Q-learning.</strong> Q(s, a) = W[a]·x + b[a], a linear function of the 47 features. This is the minimal step beyond a table and the natural bridge from Config A's tabular Q-learning: fast, transparent, and a direct measure of how much of the problem is linearly separable.</li>
        <li style="margin-right: 20px;"><strong>B.2 — DQN.</strong> The same bootstrapped Q-learning target, but Q is a neural network trained with experience replay and a target network. It can capture non-linear interactions between physiological variables that the linear model cannot.</li>
    </ul>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Comparing the two answers Config B's central question: does the continuous problem actually require a non-linear approximator, or do linear features suffice?</p>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">All Config B agents use γ = 1.0 (matching Config A - episodes terminate, so no time discounting), train for the same number of environment steps across the same three seeds (<code>SEEDS_B</code>), and are scored with the single fixed <code>evaluate_agent</code> function. Each is compared against the clinical random baseline established in <strong>1. Explore the Environment</strong>.</p>
</div>

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">B.1 — Linear Q-learning (linear function approximation)</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Semi-gradient Q-learning with linear features: for each action a we keep a weight vector W[a] (47 values) and a bias b[a], so Q(s, a) = W[a]·x + b[a]. After each transition (x, a, r, x') we form the bootstrapped target y = r + γ·max over a' of Q(x', a') (just r at a terminal step) and take a semi-gradient step toward it. Exploration uses an ε-greedy schedule decaying linearly over the first 80% of training, mirroring Config A. We train one model per seed and evaluate a fixed-seed snapshot every <code>SNAPSHOT_EVERY</code> steps to trace the learning curve.</p>
</div>

```python
def linear_q_train(seed, total_steps=TOTAL_STEPS, alpha=0.01, gamma=GAMMA,
                   eps_start=1.0, eps_end=0.05, snapshot_every=SNAP_EVERY_B,
                   eval_episodes=200):
    rng = np.random.default_rng(seed)
    env = make_clinical_env_quiet()
    # Small random init breaks the all-zero argmax tie that otherwise pins
    # the policy to action 0 forever (all 25 Q-values start identical, so
    # np.argmax always returns 0, only head 0 gets updated, and the agent
    # never explores enough to discover that any other action could be better).
    W = rng.normal(loc=0.0, scale=0.01, size=(N_ACTIONS, 47)).astype(np.float64)
    b = rng.normal(loc=0.0, scale=0.01, size=N_ACTIONS).astype(np.float64)
    eps_decay_steps = int(0.8 * total_steps)

    def greedy_action(obs_vec):
        q_values = W @ obs_vec + b
        action = int(np.argmax(q_values))
        return action

    snapshot_steps = []
    snapshot_returns = []
    eps_trace = []
    best_return = -np.inf
    best_W = W.copy()
    best_b = b.copy()
    step_count = 0
    obs, info = env.reset(seed=int(rng.integers(0, 100_000)))
    while step_count < total_steps:
        frac = min(1.0, step_count / eps_decay_steps)
        eps = eps_start + frac * (eps_end - eps_start)
        explore = rng.random() < eps
        if explore:
            action = int(rng.integers(0, N_ACTIONS))
        else:
            action = greedy_action(obs)
        next_obs, r, te, tr, info = env.step(action)
        q_sa = float(W[action] @ obs + b[action])
        terminal = te or tr
        if terminal:
            target = r
        else:
            q_next = W @ next_obs + b
            target = r + gamma * float(np.max(q_next))
        td_error = target - q_sa
        W[action] += alpha * td_error * obs
        b[action] += alpha * td_error
        step_count += 1
        if terminal:
            obs, info = env.reset(seed=int(rng.integers(0, 100_000)))
        else:
            obs = next_obs
        if step_count % snapshot_every == 0:
            saved_state = np.random.get_state()
            returns_snap, _, _ = evaluate_agent(greedy_action, n_episodes=eval_episodes, seed=SEED)
            np.random.set_state(saved_state)
            snap_mean = float(returns_snap.mean())
            snapshot_steps.append(step_count)
            snapshot_returns.append(snap_mean)
            eps_trace.append(eps)
            if snap_mean > best_return:
                best_return = snap_mean
                best_W = W.copy()
                best_b = b.copy()
    env.close()
    run = {
        'W': best_W,
        'b': best_b,
        'snapshot_steps': np.array(snapshot_steps),
        'snapshot_returns': np.array(snapshot_returns),
        'eps_trace': np.array(eps_trace),
        'seed': seed,
    }
    return run


linq_runs = []
for s in SEEDS_B:
    set_global_seeds(s)
    run = linear_q_train(s)
    linq_runs.append(run)
    print(f'Linear-Q seed {s}: final snapshot return = {run["snapshot_returns"][-1]:.4f}')

linq_final_list = []
for run in linq_runs:
    linq_final_list.append(float(np.max(run['snapshot_returns'])))
linq_best_idx = int(np.argmax(np.array(linq_final_list)))
W_linq = linq_runs[linq_best_idx]['W']
b_linq = linq_runs[linq_best_idx]['b']


def linq_predict(obs):
    q_values = W_linq @ obs + b_linq
    action = int(np.argmax(q_values))
    return action


set_global_seeds(SEED)
linq_returns, linq_lengths, linq_eval_info = evaluate_agent(linq_predict, n_episodes=1000, seed=SEED)
linq_mean_ret = float(linq_returns.mean())
linq_surv = float((linq_returns > 0).mean()) * 100
print(f'Linear-Q (best seed {SEEDS_B[linq_best_idx]}): mean return = {linq_mean_ret:.4f}, survival = {linq_surv:.1f}%')
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">B.1b - Random Fourier Feature Linear-Q (richer linear approximator)</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">B.1 showed Linear-Q collapses to a constant policy because the linear functional form $Q(s, a) = W[a] \cdot x + b[a]$ cannot express the multi-modal optimal policy of this MDP. The failure is representational, not algorithmic. To test whether a <em>richer</em> linear approximator escapes the collapse, we replace the 47-dim raw features with a 256-dim Random Fourier Feature (RFF) embedding.</p>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">RFF features approximate an RBF kernel as $D \to \infty$. For each new feature $k$ we sample fixed random direction $\omega_k \sim \mathcal{N}(0, I)$ and phase $b_k \sim \mathcal{U}(0, 2\pi)$, then compute $\phi_k(s) = \sqrt{2/D} \cos(\omega_k \cdot x_s + b_k)$. The new approximator $Q(s, a) = W[a] \cdot \phi(s) + b[a]$ is still linear in $\phi$, so the same semi-gradient TD update applies; but $\phi$ captures non-linear interactions between the 47 original features that the naive Linear-Q cannot.</p>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">If the collapse is purely representational, RFF should escape it. If RFF also collapses, the issue runs deeper (e.g., the always-action-0 attractor is too strong even for richer features).</p>
</div>

```python
RFF_DIM = 256
RFF_BANDWIDTH = 1.0


rng_rff = np.random.default_rng(42)
W_rff = rng_rff.normal(0, 1.0 / RFF_BANDWIDTH, (RFF_DIM, 47))
b_rff = rng_rff.uniform(0, 2 * np.pi, RFF_DIM)


def rff_features(obs_vec):
    proj = W_rff @ obs_vec + b_rff
    feat = np.sqrt(2.0 / RFF_DIM) * np.cos(proj)
    return feat


def linear_q_rff_train(seed, total_steps=TOTAL_STEPS, alpha=0.005, gamma=GAMMA,
                       eps_start=1.0, eps_end=0.05, snapshot_every=SNAP_EVERY_B,
                       eval_episodes=200):
    rng = np.random.default_rng(seed)
    env = make_clinical_env_quiet()
    W = rng.normal(0, 0.01, (N_ACTIONS, RFF_DIM)).astype(np.float64)
    b = rng.normal(0, 0.01, N_ACTIONS).astype(np.float64)
    eps_decay_steps = int(0.8 * total_steps)

    def greedy_action(obs_vec):
        x_rff = rff_features(obs_vec)
        q_values = W @ x_rff + b
        action = int(np.argmax(q_values))
        return action

    snapshot_steps = []
    snapshot_returns = []
    eps_trace = []
    best_return = -np.inf
    best_W = W.copy()
    best_b = b.copy()
    step_count = 0
    obs, info = env.reset(seed=int(rng.integers(0, 100_000)))
    while step_count < total_steps:
        frac = min(1.0, step_count / eps_decay_steps)
        eps = eps_start + frac * (eps_end - eps_start)
        explore = rng.random() < eps
        if explore:
            action = int(rng.integers(0, N_ACTIONS))
        else:
            action = greedy_action(obs)
        next_obs, r, te, tr, info = env.step(action)
        x_rff = rff_features(obs)
        q_sa = float(W[action] @ x_rff + b[action])
        terminal = te or tr
        if terminal:
            target = r
        else:
            x_rff_next = rff_features(next_obs)
            q_next = W @ x_rff_next + b
            target = r + gamma * float(np.max(q_next))
        td_error = target - q_sa
        W[action] += alpha * td_error * x_rff
        b[action] += alpha * td_error
        step_count += 1
        if terminal:
            obs, info = env.reset(seed=int(rng.integers(0, 100_000)))
        else:
            obs = next_obs
        if step_count % snapshot_every == 0:
            saved_state = np.random.get_state()
            returns_snap, _, _ = evaluate_agent(greedy_action, n_episodes=eval_episodes, seed=SEED)
            np.random.set_state(saved_state)
            snap_mean = float(returns_snap.mean())
            snapshot_steps.append(step_count)
            snapshot_returns.append(snap_mean)
            eps_trace.append(eps)
            if snap_mean > best_return:
                best_return = snap_mean
                best_W = W.copy()
                best_b = b.copy()
    env.close()
    run = {
        'W': best_W,
        'b': best_b,
        'snapshot_steps': np.array(snapshot_steps),
        'snapshot_returns': np.array(snapshot_returns),
        'eps_trace': np.array(eps_trace),
        'seed': seed,
    }
    return run


linq_rff_runs = []
for s in SEEDS_B:
    set_global_seeds(s)
    run = linear_q_rff_train(s)
    linq_rff_runs.append(run)
    print(f'Linear-Q-RFF seed {s}: final snapshot return = {run["snapshot_returns"][-1]:.4f}')

linq_rff_final_list = []
for run in linq_rff_runs:
    linq_rff_final_list.append(float(np.max(run['snapshot_returns'])))
linq_rff_best_idx = int(np.argmax(np.array(linq_rff_final_list)))
W_linq_rff = linq_rff_runs[linq_rff_best_idx]['W']
b_linq_rff = linq_rff_runs[linq_rff_best_idx]['b']


def linq_rff_predict(obs):
    x_rff = rff_features(obs)
    q_values = W_linq_rff @ x_rff + b_linq_rff
    action = int(np.argmax(q_values))
    return action


set_global_seeds(SEED)
linq_rff_returns, linq_rff_lengths, linq_rff_eval_info = evaluate_agent(linq_rff_predict, n_episodes=1000, seed=SEED)
linq_rff_mean_ret = float(linq_rff_returns.mean())
linq_rff_surv = float((linq_rff_returns > 0).mean()) * 100
print(f'Linear-Q-RFF (best seed {SEEDS_B[linq_rff_best_idx]}): mean return = {linq_rff_mean_ret:.4f}, survival = {linq_rff_surv:.1f}%')
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">B.2 — DQN (deep function approximation)</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">DQN keeps the Q-learning target of B.1 but represents Q(s, a) with a multi-layer perceptron (<code>MlpPolicy</code>), stabilised by an experience-replay buffer and a periodically updated target network. We use the stable-baselines3 implementation. A shared snapshot callback evaluates the greedy policy with the fixed <code>evaluate_agent</code> function at the same cadence as Linear-Q, so the learning curves are directly comparable on a common environment-steps axis.</p>
</div>

<div style="background-color: #EAF1F8; padding: 10px; margin-top: 10px; margin-right: 30px; border-left: 5px solid #4682B4; border-radius: 5px; max-width: 95%; color: #333333;">
    <strong>Hyperparameter choices (tuned from SB3 defaults):</strong>
    <ul style="margin: 6px 40px 0 20px; padding-left: 20px; color: #333333;">
        <li><code>learning_rate = 1e-4</code> &mdash; five times smaller than the SB3 default for MlpPolicy DQN. The smaller step size makes TD updates smoother and prevents the oscillations we saw at the default $5 \times 10^{-4}$.</li>
        <li><code>tau = 0.005</code>, <code>target_update_interval = 1</code> &mdash; soft target updates replace the hard copy every 1{,}000 steps. The target network moves slowly toward the online network every step rather than jumping abruptly, removing a major source of training instability.</li>
        <li><code>buffer_size = 200{,}000</code> &mdash; covers the full 300{,}000-step training horizon with margin; the replay distribution is non-stationary only on the policy axis.</li>
        <li><code>batch_size = 128</code> &mdash; larger batches reduce gradient variance for the small-MLP architecture; combined with the smaller learning rate, the per-update signal-to-noise improves.</li>
        <li><code>train_freq = 4</code>, <code>gradient_steps = 1</code> &mdash; one gradient step per four environment interactions. Standard DQN ratio.</li>
        <li><code>total_timesteps = 300{,}000</code> &mdash; double our original Linear-Q budget. The smaller learning rate needs more steps to converge.</li>
        <li><code>exploration_fraction = 0.2</code>, <code>exploration_final_eps = 0.05</code> &mdash; unchanged from the original setup, matches Configuration A.</li>
    </ul>
</div>

```python
class EvalSnapshotCallback(BaseCallback):
    def __init__(self, eval_every=SNAP_EVERY_B, eval_episodes=200, eval_seed=SEED):
        super().__init__()
        self.eval_every = eval_every
        self.eval_episodes = eval_episodes
        self.eval_seed = eval_seed
        self.snapshot_steps = []
        self.snapshot_returns = []
        self.eps_trace = []
        self._last_eval = 0
        self.best_return = -np.inf
        self.best_params = None

    def _model_predict(self, obs):
        action, _ = self.model.predict(obs, deterministic=True)
        return int(action)

    def _on_step(self):
        due = (self.num_timesteps - self._last_eval) >= self.eval_every
        if due:
            self._last_eval = self.num_timesteps
            saved_state = np.random.get_state()
            returns_snap, _, _ = evaluate_agent(self._model_predict,
                                                n_episodes=self.eval_episodes, seed=self.eval_seed)
            np.random.set_state(saved_state)
            snap_mean = float(returns_snap.mean())
            self.snapshot_steps.append(self.num_timesteps)
            self.snapshot_returns.append(snap_mean)
            current_eps = getattr(self.model, 'exploration_rate', np.nan)
            self.eps_trace.append(current_eps)
            if snap_mean > self.best_return:
                self.best_return = snap_mean
                self.best_params = copy.deepcopy(self.model.get_parameters())
        keep_going = True
        return keep_going

def train_sb3(make_model, label, seeds=SEEDS_B, total_steps=TOTAL_STEPS):
    runs = []
    for s in seeds:
        set_global_seeds(s)
        train_env = Monitor(make_clinical_env_quiet())
        model = make_model(train_env, s)
        callback = EvalSnapshotCallback()
        model.learn(total_timesteps=total_steps, callback=callback)
        if callback.best_params is not None:
            model.set_parameters(callback.best_params)
        train_env.close()
        run = {
            'model': model,
            'snapshot_steps': np.array(callback.snapshot_steps),
            'snapshot_returns': np.array(callback.snapshot_returns),
            'eps_trace': np.array(callback.eps_trace),
            'seed': s,
        }
        runs.append(run)
        print(f'{label} seed {s}: final snapshot return = {run["snapshot_returns"][-1]:.4f}')
    return runs

def sb3_predict_factory(model):
    def predict(obs):
        action, _ = model.predict(obs, deterministic=True)
        return int(action)
    return predict

DQN_STEPS = 300_000
DQN_SNAP_EVERY = 20_000


def make_dqn(env, seed):
    model = DQN(
        'MlpPolicy', env,
        gamma=GAMMA,
        learning_rate=1e-4,        # 5x smaller than SB3 default; smoother TD updates
        buffer_size=200_000,        # covers the full 300k horizon without overwrite
        learning_starts=2_000,      # extra warm-up
        batch_size=128,             # larger batch reduces gradient variance
        tau=0.005,                  # soft target updates instead of hard copies
        target_update_interval=1,   # required by SB3 for soft tau
        train_freq=4,               # one gradient step per 4 env steps
        gradient_steps=1,
        exploration_fraction=0.2,
        exploration_final_eps=0.05,
        seed=seed,
        verbose=0,
    )
    return model


def train_sb3_custom(make_model, label, seeds=SEEDS_B, total_steps=DQN_STEPS, snap_every=DQN_SNAP_EVERY):
    runs = []
    for s in seeds:
        set_global_seeds(s)
        train_env = Monitor(make_clinical_env_quiet())
        model = make_model(train_env, s)
        callback = EvalSnapshotCallback(eval_every=snap_every)
        model.learn(total_timesteps=total_steps, callback=callback)
        if callback.best_params is not None:
            model.set_parameters(callback.best_params)
        train_env.close()
        run = {
            'model': model,
            'snapshot_steps': np.array(callback.snapshot_steps),
            'snapshot_returns': np.array(callback.snapshot_returns),
            'eps_trace': np.array(callback.eps_trace),
            'seed': s,
        }
        runs.append(run)
        print(f'{label} seed {s}: final snapshot return = {run["snapshot_returns"][-1]:.4f}')
    return runs


dqn_runs = train_sb3_custom(make_dqn, 'DQN', total_steps=DQN_STEPS, snap_every=DQN_SNAP_EVERY)
dqn_best = best_run(dqn_runs)
dqn_model = dqn_best['model']
dqn_predict = sb3_predict_factory(dqn_model)

set_global_seeds(SEED)
dqn_returns, dqn_lengths, dqn_eval_info = evaluate_agent(dqn_predict, n_episodes=1000, seed=SEED)
dqn_mean_ret = float(dqn_returns.mean())
dqn_surv = float((dqn_returns > 0).mean()) * 100
print(f'DQN (best seed {dqn_best["seed"]}): mean return = {dqn_mean_ret:.4f}, survival = {dqn_surv:.1f}%')
```

## <div style="display:inline-block; background:linear-gradient(to right, #4682B4, #2C5F8A, #1A3D5C); color:#FFFFFF; font-weight:bold; font-size:18px; padding:7px 18px; border-radius:6px; font-family:'Roboto','Segoe UI','Calibri','Open Sans',sans-serif;">B.3 — Config B results and interpretation</div>

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">

<div style="background-color: #F2F2F2; padding: 15px; margin-right: 30px; border-left: 5px solid; border-image: linear-gradient(to bottom, #A8C5E0, #4682B4, #2C5F8A, #1A3D5C) 1; border-radius: 5px; max-width: 95%;">
    <h3 style="margin-top: 0; margin-bottom: 10px; color: #333333; font-weight: bold;">Motivation</h3>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Every algorithm so far optimises the expected return. In a clinical context that objective is uncomfortable: a policy that gets +1 in 78% of patients and 0 in 22% has the same expected value as one that gets +0.78 in every patient. The first kills 22 patients per 100; the second kills none. The mean cannot tell them apart.</p>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">The extension replaces the risk-neutral objective with a risk-sensitive one and asks what the policy then does. The implementation is a Lagrangian shaping of the terminal reward: at the death transition the agent receives <code>r = 0 - β</code> instead of 0, while at the survival transition and all intermediate transitions the reward is unchanged. For β = 0 the algorithm reduces exactly to standard Q-Learning; for β &gt; 0 the same Bellman recursion solves a different optimisation problem.</p>
    <h4 style="margin-top: 15px; margin-bottom: 8px; color: #4682B4; font-weight: bold;">Why this is the right move</h4>
    <ul style="margin: 10px 40px 10px 20px; padding-left: 20px; padding-right: 0; color: #757575;">
        <li style="margin-right: 20px;"><strong>Lagrangian view.</strong> For any β &gt; 0, the β-shifted optimum lies on the Pareto frontier of "maximise expected non-death reward subject to P(death | π) ≤ ε". Sweeping β traces out that frontier without ever changing the algorithm.</li>
        <li style="margin-right: 20px;"><strong>Drops cleanly into existing infrastructure.</strong> The change is one branch in the terminal target: if <code>next_state == STATE_DIED</code>, subtract β. Evaluation still uses the true reward (no β), so survival and the original return distribution are read without contamination.</li>
        <li style="margin-right: 20px;"><strong>Reads as a clinical knob.</strong> β is the "currency" of an avoided death relative to a survival reward of +1. β = 1 says "one death equals one foregone survival"; β = 2 says "one death is worth two foregone survivals". Stakeholders can argue about its value; the algorithm just computes the resulting policy.</li>
    </ul>
    <h4 style="margin-top: 15px; margin-bottom: 8px; color: #4682B4; font-weight: bold;">Sweep design</h4>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">β ∈ {0, 0.5, 1.0, 2.0}; one seed; 100,000 episodes per β; α = 0.1; the same ε-decay 1.0 → 0.05 schedule as A.2. We report survival rate, mean return, 5%-quantile return (worst-case proxy), and mean episode length under each β, plus a heatmap of the β = 2 policy contrasted with the standard QL policy at β = 0.</p>
</div>

```python
RISK_EPISODES = 100_000
RISK_SEEDS = [0, 1, 2, 3, 4]
RISK_BETAS = [0.0, 0.5, 1.0, 2.0]
RISK_EPS_END = 0.05


def train_risk_sensitive_ql(seed, beta, n_episodes=RISK_EPISODES, alpha=ALPHA):
    env_train = make_sepsis_env(quiet=True)
    env_train.action_space.seed(seed)
    rng = np.random.default_rng(seed)
    Q = np.zeros((N_STATES, N_ACTIONS))
    eps_decay_end = int(0.8 * n_episodes)
    for ep in range(n_episodes):
        frac = min(ep / eps_decay_end, 1.0)
        eps = 1.0 + frac * (RISK_EPS_END - 1.0)
        obs, _ = env_train.reset(seed=int(rng.integers(0, 100_000)))
        done = False
        while not done:
            if rng.random() < eps:
                action = int(env_train.action_space.sample())
            else:
                action = int(Q[obs].argmax())
            next_obs, r, te, tr, _ = env_train.step(action)
            terminal = te or tr
            if terminal:
                if int(next_obs) == STATE_DIED:
                    effective_r = r - beta
                else:
                    effective_r = r
                target = effective_r
            else:
                target = r + GAMMA * Q[next_obs].max()
            td_error = target - Q[obs, action]
            Q[obs, action] += alpha * td_error
            obs = next_obs
            done = terminal
    env_train.close()
    return Q


risk_results = []
for beta_value in RISK_BETAS:
    seed_records = []
    for seed_value in RISK_SEEDS:
        Q_seed_beta = train_risk_sensitive_ql(seed_value, beta_value)
        pi_seed_beta = Q_seed_beta.argmax(axis=1)
        returns_seed_beta, lengths_seed_beta = evaluate_policy(pi_seed_beta)
        surv_seed_beta = float((returns_seed_beta > 0).mean()) * 100
        mean_seed_beta = float(returns_seed_beta.mean())
        q05_seed_beta = float(np.quantile(returns_seed_beta, 0.05))
        J_seed_beta, _ = analytical_J(pi_seed_beta)
        len_seed_beta = float(lengths_seed_beta.mean())
        seed_record = {
            'seed': seed_value, 'Q': Q_seed_beta, 'pi': pi_seed_beta,
            'returns': returns_seed_beta, 'lengths': lengths_seed_beta,
            'surv': surv_seed_beta, 'mean': mean_seed_beta, 'q05': q05_seed_beta,
            'J': J_seed_beta, 'mean_len': len_seed_beta,
        }
        seed_records.append(seed_record)
    surv_arr = np.array([rec['surv'] for rec in seed_records])
    mean_arr = np.array([rec['mean'] for rec in seed_records])
    q05_arr = np.array([rec['q05'] for rec in seed_records])
    J_arr = np.array([rec['J'] for rec in seed_records])
    len_arr = np.array([rec['mean_len'] for rec in seed_records])
    aggregate = {
        'beta': beta_value,
        'seeds': seed_records,
        'surv_mean': float(surv_arr.mean()), 'surv_std': float(surv_arr.std()),
        'mean_mean': float(mean_arr.mean()), 'mean_std': float(mean_arr.std()),
        'q05_mean': float(q05_arr.mean()), 'q05_std': float(q05_arr.std()),
        'J_mean': float(J_arr.mean()), 'J_std': float(J_arr.std()),
        'len_mean': float(len_arr.mean()), 'len_std': float(len_arr.std()),
    }
    risk_results.append(aggregate)
    print(f'beta={beta_value:.1f} | surv={aggregate["surv_mean"]:5.1f}% +/- {aggregate["surv_std"]:.1f} '
          f'| mean={aggregate["mean_mean"]:.4f} +/- {aggregate["mean_std"]:.4f} '
          f'| 5%-q={aggregate["q05_mean"]:.4f} +/- {aggregate["q05_std"]:.4f} '
          f'| J(pi)={aggregate["J_mean"]:.4f} +/- {aggregate["J_std"]:.4f}')
```

    <h3 style="margin-top: 18px; margin-bottom: 10px; color: #333333; font-weight: bold;">Interpretation</h3>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Two findings come out of this. First, the &lambda;-intensity penalty is doing more work than expected: it suppresses treatment so aggressively that even the standard Q-Learning policy lands below the survival-maximising frontier. Adding a death penalty pushes the agent back toward "treat more, even if it costs intensity." Second, the multi-seed sweep shows &beta;=0.5 as a clinically attractive sweet spot: highest mean return and survival within 1 pp of the &beta;=2 peak. The single-seed result that suggested monotonic mean growth was misleading.</p>
    <h3 style="margin-top: 18px; margin-bottom: 10px; color: #333333; font-weight: bold;">What this adds</h3>
    <p style="margin: 10px 0; margin-right: 40px; color: #757575;">Reframes the objective rather than the approximator. Every other algorithm in this notebook (VI, QL, SARSA, Linear-Q, DQN) optimises expectation; &beta; &gt; 0 places the policy on a different point of the survival / intensity trade-off without changing the environment, the action space, or the algorithm class. It is the same Bellman recursion solving a different question, and the multi-seed result surfaces a non-trivial Pareto frontier with a real sweet spot at &beta;=0.5.</p>

    <h3 style="margin-top: 0; margin-bottom: 10px; color: #333333; font-weight: bold;">Limitations and future work</h3>
    <ul style="margin: 10px 40px 10px 20px; padding-left: 20px; padding-right: 0; color: #757575;">
        <li style="margin-right: 20px;"><strong>Simulator, not patients.</strong> Every result here is on the ICU-Sepsis-v2 MDP, which is fitted to the MIMIC-III cohort. Policies that look good in simulation can still be wrong on real patients whose state-transition dynamics differ from the fitted MDP.</li>
        <li style="margin-right: 20px;"><strong>Three seeds.</strong> Cross-seed std is informative but underpowered for tight statistical claims. The QL vs SARSA "at parity" conclusion would benefit from 10+ seeds; the same applies to the risk-sensitive sweep where one seed per &beta; could mask sensitivity.</li>
        <li style="margin-right: 20px;"><strong>DQN never converged.</strong> 150k steps with the chosen hyperparameters leave the curve oscillating. A longer budget, a smaller learning rate, or a soft target update (&tau; &lt; 1) might let DQN close the gap to the oracle; the current result is a lower bound on what deep RL could achieve here.</li>
        <li style="margin-right: 20px;"><strong>Single risk-sensitive seed per &beta;.</strong> The Section 5 sweep is a clean illustration of the Pareto frontier but a single training seed per &beta; cannot separate "&beta; matters" from "this particular run was lucky". Multi-seed replication is the obvious next step.</li>
        <li style="margin-right: 20px;"><strong>Clinical realism of the wrappers.</strong> The episodic-noise / missing-labs / acute-event wrappers are reasonable proxies but not calibrated to real ICU error rates. A robustness study that sweeps each wrapper's probability would tell whether the agents fail gracefully or cliff-edge.</li>
        <li style="margin-right: 20px;"><strong>Action interpretability.</strong> Vasopressor / fluid levels are intuitive, but the policy never explains <em>why</em> it chose a particular cell in a particular state. Saliency or SHAP on DQN, or coefficient-level inspection of Linear-Q, would close that gap for a clinical reader.</li>
    </ul>
